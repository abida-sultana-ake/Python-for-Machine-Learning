## 1.1 What Is NumPy?

**NumPy** stands for **Numerical Python**. It is a Python library used to work with numbers, arrays, and mathematical calculations efficiently.

Imagine L from *Death Note* is investigating several suspects. He records their investigation scores:

```python
scores = [75, 80, 90, 65, 85]
```

Now L wants to increase every suspect's score by 10 points. Doing this manually would be boring, even for L.

NumPy helps us perform calculations on an entire collection of numbers at once.

### Why Do We Need NumPy?

Suppose you are tracking the marks of five students.

Without NumPy, you may need a loop or list comprehension to update every mark.

With NumPy, you can perform the operation directly on the whole array.

**Real-life applications of NumPy:**
- Calculating students' marks and averages.
- Analyzing traffic speed and sensor data.
- Processing image data.
- Performing mathematical calculations in AI and Machine Learning.

---

## 1.2 Python List vs NumPy Array

### A. Using a Python List

```python
scores = [75, 80, 90, 65, 85]

print(scores)
```

**Output:**

```text
[75, 80, 90, 65, 85]
```

Now imagine L wants to add 10 points to every score.

```python
scores = [75, 80, 90, 65, 85]

print(scores + 10)
```

**Output:**

```text
TypeError
```

Why?

Python does not allow us to add a single integer directly to an entire list.

L must investigate every score individually!

We can use a list comprehension:

```python
scores = [75, 80, 90, 65, 85]

updated_scores = [score + 10 for score in scores]

print(updated_scores)
```

**Output:**

```text
[85, 90, 100, 75, 95]
```

It works, but NumPy provides a more convenient way to perform numerical operations.

### B. Using a NumPy Array

First, import NumPy:

```python
import numpy as np
```

Here:
- `import` loads a library.
- `numpy` is the library's name.
- `as np` gives it a shorter nickname.

Think of `np` as L's secret investigation code name.

Now create a NumPy array:

```python
import numpy as np

scores = np.array([75, 80, 90, 65, 85])

print(scores)
```

**Output:**

```text
[75 80 90 65 85]
```

Notice that NumPy displays the numbers without commas.

Now increase every score by 10:

```python
updated_scores = scores + 10

print(updated_scores)
```

**Output:**

```text
[ 85  90 100  75  95]
```

NumPy adds 10 to every element automatically!

This is called an **element-wise operation**.

```python
scores = np.array([10, 20, 30])

print(scores + 5)
print(scores * 2)
print(scores - 3)
```

**Output:**

```text
[15 25 35]
[20 40 60]
[ 7 17 27]
```

**Remember:** When we perform these arithmetic operations on a NumPy array, NumPy applies the operation to each element.

`ndarray` means **N-dimensional array**. It is the main array object provided by NumPy.

In simple words, whenever we create a NumPy array, we usually create an `ndarray` object.

```python
import numpy as np

suspect_scores = np.array([75, 80, 90, 65, 85])

print(suspect_scores)
print(type(suspect_scores))
```

**Output:**

```text
[75 80 90 65 85]
<class 'numpy.ndarray'>
```

### Understanding the Code

- `np.array()` creates a NumPy array.
- `type()` tells us the type of an object.
- `numpy.ndarray` tells us that `suspect_scores` is a NumPy array.

**L's rule:** Before investigating the suspects, identify what kind of evidence you have!

## 2.2 What Are Array Attributes?

Attributes provide information about an array.

Think of them as L's investigation report. They tell him how the evidence is organized, how many elements exist, and what type of data is stored.

We will learn five useful attributes:

| Attribute | What it tells us |
|---|---|
| `ndim` | Number of dimensions |
| `shape` | Size of the array along each dimension |
| `size` | Total number of elements |
| `dtype` | Data type of the elements |
| `itemsize` | Memory used by each element, in bytes |

Remember: **Attributes usually do not need parentheses.** For example, use `arr.shape`, not `arr.shape()`.

`shape` tells us the size of an array along each dimension.

It means the array has **5 elements in one dimension**.

The comma is part of how Python represents a one-element tuple.

This means:

- `3` rows
- `3` columns

## 2.6 What Is a Data Type (`dtype`)?

A **data type** describes what kind of value is stored in an array.

For example:

- Integers: `10`, `25`, `100`
- Floating-point numbers: `10.5`, `25.75`
- Boolean values: `True`, `False`

NumPy uses the `dtype` attribute to report an array's element type.

### Example 1: Integer Array

```python
scores = np.array([75, 80, 90])

print(scores.dtype)
```

**Possible output:**

```text
int64
```

On many 64-bit systems, NumPy uses `int64` for this array. The exact default integer type can depend on the platform.

### Example 2: Floating-Point Array

```python
prices = np.array([99.5, 150.75, 200.0])

print(prices.dtype)
```

**Common output:**

```text
float64
```

These values contain decimal points, so NumPy typically creates a floating-point array.

### Example 3: Boolean Array

```python
suspect_found = np.array([True, False, True])

print(suspect_found.dtype)
```

**Output:**

```text
bool
```

Here, `bool` represents Boolean data.

### Example 4: Text Data

```python
suspects = np.array(["Light", "L", "Near"])

print(suspects.dtype)
```

NumPy creates a string array. Its exact displayed dtype depends on the string representation used by the installed NumPy version.

**Important:** A standard NumPy array generally stores elements using a common data type. When an array contains mixed types, NumPy may convert the values to a compatible common type.

---

## 2.7 What Is `itemsize`?

`itemsize` tells us how many bytes are used by each array element.

```python
numbers = np.array([10, 20, 30])

print(numbers.dtype)
print(numbers.itemsize)
```

**Possible output on a 64-bit system:**

```text
int64
8
```

An `int64` element uses 64 bits, and 8 bits make 1 byte:

\(64 \div 8 = 8\) bytes.

If an array contains 3 elements and each element uses 8 bytes, the element storage is:

\(3 \times 8 = 24\) bytes.

You can check the total element storage with `nbytes`:

```python
print(numbers.nbytes)
```

**Output:**

```text
24
```

The exact result depends on the array's data type and number of elements.

For now, remember:

- `itemsize`: bytes per element.
- `nbytes`: total bytes used by the array's elements.

This does not include every bit of overhead associated with the NumPy array object.

---

## 2.8 Choosing a Data Type with `dtype=`

Sometimes we want to choose the data type ourselves.

We can do that using the `dtype` parameter inside `np.array()`.

### Example 1: Choose an Integer Type

```python
numbers = np.array([10, 20, 30], dtype=np.int32)

print(numbers)
print(numbers.dtype)
print(numbers.itemsize)
```

**Output:**

```text
[10 20 30]
int32
4
```

`int32` uses 32 bits, which is 4 bytes per element.

### Example 2: Choose a Floating-Point Type

```python
prices = np.array([10, 20, 30], dtype=np.float32)

print(prices)
print(prices.dtype)
```

**Output:**

```text
[10. 20. 30.]
float32
```

Even though we supplied integers, NumPy stores them as floating-point values.

### Example 3: Choose Boolean Type

```python
results = np.array([1, 0, 1], dtype=np.bool_)

print(results)
print(results.dtype)
```

**Output:**

```text
[ True False  True]
bool
```

Here, `1` becomes `True` and `0` becomes `False`.

### Common NumPy Data Types

| Data type | Meaning | Example |
|---|---|---|
| `np.int32` | 32-bit integer | `25` |
| `np.int64` | 64-bit integer | `250000` |
| `np.float32` | 32-bit floating-point number | `3.14` |
| `np.float64` | 64-bit floating-point number | `99.99` |
| `np.bool_` | Boolean value | `True` |

You do not need to memorize every data type today. Understand what they mean and how to select one.

## 2.9 Full Investigation Report

Let's combine the attributes in one example.

## 3.1 Creating Arrays from Existing Data

We have already learned `np.array()`. It creates a NumPy array from existing data, such as a Python list or tuple.

### Example 1: From a List

```python
import numpy as np

scores = [75, 80, 90, 65]

score_array = np.array(scores)

```

**Output:**


## 3.4 Creating Arrays with Range Functions

Sometimes we want a sequence of numbers rather than randomly generated values.

For example, L wants to investigate scores from 0 to 10 or create evenly spaced values for a graph.

Two useful functions are `np.arange()` and `np.linspace()`.

### A. `np.arange()`

Creates a sequence of numbers using a start, stop, and optional step.

**Syntax:**

```python
np.arange(start, stop, step)
```

The `stop` value is excluded.

#### Example 1: Start and Stop

```python
numbers = np.arange(1, 6)

print(numbers)
```

**Output:**

```text
[1 2 3 4 5]
```

It starts at 1 and stops before 6.

#### Example 2: Use a Step

```python
numbers = np.arange(0, 10, 2)

print(numbers)
```

**Output:**

```text
[0 2 4 6 8]
```

Here:

- Start = `0`
- Stop = `10` (excluded)
- Step = `2`

#### Example 3: Decimal Steps

```python
numbers = np.arange(0, 1, 0.2)

print(numbers)
```

This creates values stepping by approximately `0.2` from 0, stopping before 1. Floating-point calculations can sometimes produce small precision differences.

### B. `np.linspace()`

Creates a specified number of evenly spaced values between a start and stop value.

**Syntax:**

```python
np.linspace(start, stop, num)
```

By default, both endpoints are included.

#### Example 1: Five Values

```python
numbers = np.linspace(0, 10, 5)

print(numbers)
```

**Output:**

```text
[ 0.   2.5  5.   7.5 10. ]
```

NumPy divides the interval from 0 to 10 into four equal gaps to create five values.

#### Example 2: Three Values

```python
numbers = np.linspace(0, 100, 3)

print(numbers)
```

**Output:**

```text
[  0.  50. 100.]
```

We asked for three values, so NumPy returned three evenly spaced numbers including both endpoints.

### `np.arange()` vs `np.linspace()`

| Feature | `np.arange()` | `np.linspace()` |
|---|---|---|
| Main control | Step size | Number of values |
| Stop value | Excluded | Included by default |
| Example | `np.arange(0, 10, 2)` | `np.linspace(0, 10, 5)` |
| Output | `[0, 2, 4, 6, 8]` | `[0., 2.5, 5., 7.5, 10.]` |

**Memory trick:**

- **`arange` = arrange numbers by a step.**
- **`linspace` = line up a specific number of evenly spaced values.**

### Real-Life Example: Traffic Monitoring

A traffic sensor records speed values from 0 to 100 km/h.

```python
speed_points = np.linspace(0, 100, 6)

print(speed_points)
```

**Output:**

```text
[  0.  20.  40.  60.  80. 100.]
```

This produces six evenly spaced speed points that could be useful for plotting or testing.

# Lesson 4: NumPy Array Indexing and Slicing
### A Beginner's Friendly Survival Guide

Imagine L has collected evidence from five suspects. He does not always need to investigate everyone. Sometimes he needs only one suspect's score, sometimes the first three suspects, and sometimes the last suspect's record.

**Indexing** helps us select specific elements. **Slicing** helps us select a range of elements.

Let's investigate!

---

## 4.1 What Is Indexing?

**Indexing** means accessing a specific element from an array using its position.

In Python and NumPy, indexing starts at **0**, not 1.

Look at this array:

```python
import numpy as np

scores = np.array([75, 80, 90, 65, 85])

print(scores)
```

**Output:**

```text
[75 80 90 65 85]
```

Imagine that L assigns each suspect a position:

| Suspect's position | Index | Score |
|---|---:|---:|
| Suspect 1 | 0 | 75 |
| Suspect 2 | 1 | 80 |
| Suspect 3 | 2 | 90 |
| Suspect 4 | 3 | 65 |
| Suspect 5 | 4 | 85 |

Notice that the first element has index `0`.

L may be a genius detective, but even he has to follow Python's indexing rules!

## 4.2 Accessing Elements Using Indexing

We use square brackets `[]` to access an element.

### Example 1: Get the First Element

```python
scores = np.array([75, 80, 90, 65, 85])

print(scores[0])
```

**Output:**

```text
75
```

`scores[0]` means: Give me the element at index 0.

### Example 2: Get the Third Element

```python
print(scores[2])
```

**Output:**

```text
90
```

Why `2`, not `3`? Because indexing starts from zero.

### Example 3: Get the Last Element Using a Positive Index

```python
print(scores[4])
```

**Output:**

```text
85
```

The array has five elements, so its indexes are `0, 1, 2, 3, 4`.

### Example 4: Update an Element

L discovered that a suspect's score was recorded incorrectly. He needs to correct it.

```python
scores = np.array([75, 80, 90, 65, 85])

scores[3] = 95

print(scores)
```

**Output:**

```text
[75 80 90 95 85]
```

We changed the element at index `3` from `65` to `95`.

**Remember:** Indexing can be used to read an element or update an existing element.

---

## 4.3 Negative Indexing

What if L wants the last suspect's score but does not want to count all the indexes?

NumPy supports negative indexing.

Negative indexing starts from the end of an array.

| Index | Meaning |
|---:|---|
| `-1` | Last element |
| `-2` | Second-last element |
| `-3` | Third-last element |

Example:

```python
scores = np.array([75, 80, 90, 65, 85])

print(scores[-1])
print(scores[-2])
print(scores[-3])
```

**Output:**

```text
85
65
90
```

### Why Is This Useful?

Imagine L receives a file containing 10,000 investigation scores. He wants only the most recent score.

Instead of calculating the last positive index, he can simply write:

```python
latest_score = scores[-1]
```

**L's shortcut:** When you need the last element, use `-1`.

---

## 4.4 What Is Slicing?

**Slicing** means selecting a range of elements from an array.

The basic syntax is:

```python
array[start:stop:step]
```

| Part | Meaning |
|---|---|
| `start` | Index where selection begins |
| `stop` | Index where selection ends, excluded |
| `step` | How many positions to move each time |

The `stop` index is **not included** in the result.

Don't worry if this looks confusing. Let's break it down.

### Example 1: Select the First Three Elements

```python
scores = np.array([75, 80, 90, 65, 85])

print(scores[0:3])
```

**Output:**

```text
[75 80 90]
```

The selection starts at index `0` and stops before index `3`.

So we get indexes `0`, `1`, and `2`.

### Example 2: Select Elements from Index 1 to Index 3

```python
print(scores[1:4])
```

**Output:**

```text
[80 90 65]
```

It includes indexes `1`, `2`, and `3`, but excludes index `4`.

**Memory trick:** Start is included; stop is excluded.

### Example 3: Omit the Start Index

If we omit `start`, slicing begins at the first element.

```python
print(scores[:3])
```

**Output:**

```text
[75 80 90]
```

This is equivalent to `scores[0:3]`.

### Example 4: Omit the Stop Index

If we omit `stop`, slicing continues to the end.

```python
print(scores[2:])
```

**Output:**

```text
[90 65 85]
```

This selects elements from index `2` through the end.

### Example 5: Copy the Entire Range with Slicing

```python
print(scores[:])
```

**Output:**

```text
[75 80 90 65 85]
```

Here, `[:]` selects the whole range. For a NumPy array, this normally returns a view of the original array, not an independent copy.

---

## 4.5 Using the Step in Slicing

The `step` determines how far we move between selected elements.

### Example 1: Select Every Second Element

```python
numbers = np.array([10, 20, 30, 40, 50, 60])

print(numbers[::2])
```

**Output:**

```text
[10 30 50]
```

The indexes selected are `0`, `2`, and `4`.

### Example 2: Select Every Third Element

```python
print(numbers[::3])
```

**Output:**

```text
[10 40]
```

The indexes selected are `0` and `3`.

### Example 3: Select a Range with a Step

```python
print(numbers[1:6:2])
```

**Output:**

```text
[20 40 60]
```

This starts at index `1`, stops before index `6`, and selects every second element.

### Example 4: Reverse an Array

Here's a useful shortcut:

```python
print(numbers[::-1])
```

**Output:**

```text
[60 50 40 30 20 10]
```

The step is `-1`, so NumPy moves backward through the array.

**Real-life connection:** L wants to review the investigation records in reverse order. One line of code does the job!

Now imagine L has a table of marks. Each row represents a student, and each column represents a subject.


**Output:**

```text
[[80 85 90]
 [75 88 92]
 [90 95 89]]
```

For a 2D array, we can select an element using:

```python
array[row_index, column_index]
```

The row and column indexes both start at `0`.

### Example 1: Get the First Student's First Subject Mark

```python
print(marks[0, 0])
```

**Output:**

```text
80
```

Here:

- Row index `0` selects the first student.
- Column index `0` selects the first subject.

### Example 2: Get the Second Student's Third Subject Mark

```python
print(marks[1, 2])
```

**Output:**

```text
92
```

Row `1` is the second student, and column `2` is the third subject.

### Example 3: Get the Entire Second Row

```python
print(marks[1])
```

**Output:**

```text
[75 88 92]
```

We selected row index `1`, so NumPy returned that student's complete row.

### Example 4: Get the Entire First Column

```python
print(marks[:, 0])
```

**Output:**

What does `:` mean here?

## 4.7 Slicing a Two-Dimensional Array

We can also select a range of rows and columns.

The syntax is:

```python
array[row_start:row_stop, column_start:column_stop]
```

### Example 2: Select the First Two Columns

We include every row, but only column indexes `0` and `1`.

## 4.8 Important: Slicing Can Change the Original Array

This is a very important NumPy concept.

For many basic slicing operations, NumPy returns a **view** of the original array. That means changes made through the sliced result can also affect the original array.

Example:

```python
numbers = np.array([10, 20, 30, 40, 50])

selected = numbers[1:4]

selected[0] = 999

print(selected)
print(numbers)
```

**Output:**

```text
[999  30  40]
[ 10 999  30  40  50]
```

Why did the original array change?

Because `selected` is a view that refers to the same underlying data for this basic slice.

### How Can We Avoid This?

Use `.copy()` when you need an independent array.

```python
numbers = np.array([10, 20, 30, 40, 50])

selected = numbers[1:4].copy()

selected[0] = 999

print(selected)
print(numbers)
```

**Output:**

```text
[999  30  40]
[10 20 30 40 50]
```

Now the original array remains unchanged.

**L's rule:** If you want separate evidence that can be modified independently, use `.copy()`.

### L's Final Rule

````python
# Indexing
import numpy as np

scores = np.array([75, 80, 90, 65, 85])

print(scores)
````

````python
# Accessing Elements Using Indexing
# Example 1: Get the First Element
scores = np.array([75, 80, 90, 65, 85])
print(scores[0])

# Example 2: Get the Third Element
print(scores[2])

# Example 3: Get the Last Element Using a Positive Index
print(scores[4])
print(scores[-1])

# Example 4: Update an Element
scores = np.array([75, 80, 90, 65, 85])

scores[3] = 95

print(scores)

````

````python
# Negative Indexing
scores = np.array([75, 80, 90, 65, 85])

print(scores[-1])
print(scores[-2])
print(scores[-3])
````

````python
# Slicing array[start:stop:step]
# Example 1: Select the First Three Elements
scores = np.array([75, 80, 90, 65, 85])

print(scores[0:3])

# Example 2: Select Elements from Index 1 to Index 3
print(scores[1:4])

# Example 3: Omit the Start Index If we omit start, slicing begins at the first element.
print(scores[:3])

#Example 4: Omit the Stop Index
print(scores[2:])

#Example 5: Copy the Entire Range with Slicing
print(scores[:])
````

````python
# Using the Step in Slicing
# Example 1: Select Every Second Element
numbers = np.array([10, 20, 30, 40, 50, 60])

print(numbers[::2])

# Example 2: Select Every Third Element
print(numbers[::3])

# Example 3: Select a Range with a Step
print(numbers[1:6:2])

# Example 4: Reverse an Array
print(numbers[::-1])
````

````python
# Indexing in a Two-Dimensional Array
marks = np.array([
    [80, 85, 90],
    [75, 88, 92],
    [90, 95, 89]
])

print(marks)

# Example 1: Get the First Student's First Subject Mark
print(marks[0, 0])

#Example 2: Get the Second Student's Third Subject Mark
print(marks[1, 2])

# Example 3: Get the Entire Second Row
print(marks[1])

# Example 4: Get the Entire First Column
print(marks[:, 0])

# Example 5: Get the Third Column
print(marks[:, 2])
````

````python
# Slicing a Two-Dimensional Array 

# array[row_start:row_stop, column_start:column_stop]

# Example 1: Select the First Two Rows
print(marks[0:2, :])

# Example 2: Select the First Two Columns
print(marks[:, 0:2])

# Example 3: Select a Small Section
print(marks[0:2, 1:3])
````

````python
#  Important: Slicing Can Change the Original Array

umbers = np.array([10, 20, 30, 40, 50])

selected = numbers[1:4]

selected[0] = 999

print(selected)
print(numbers) # Wrong
````

````python
# Must
numbers = np.array([10, 20, 30, 40, 50])

selected = numbers[1:4].copy()

selected[0] = 999

print(selected)
print(numbers)
````
