### 1.2 Joining Strings: Connecting the Clues

### 1.3 F-Strings: Making Interrogations Easier


Let's say Light's notebook contains a suspicious entry, and L needs to display a message containing a character's name.

You could use concatenation, but f-strings are often cleaner.

```python
suspect = "Light Yagami"
case = "Kira Investigation"

print(f"Suspect: {suspect}")
print(f"Case: {case}")
```

**Output:**

```text
Suspect: Light Yagami
Case: Kira Investigation
```

An f-string starts with `f` before the quotation mark. Variables inside `{}` are replaced with their values.

Want a more dramatic example?

```python
detective = "L"
case_status = "still suspicious"

print(f"{detective} says the case is {case_status}.")
```

**Output:**

```text
L says the case is still suspicious.
```

*Original Death Note-inspired line. Not a quote from the anime.*

### A classic mistake

```python
age = 17

# This causes a TypeError:
# print("Light's age is " + age)
```

Why? `"Light's age is "` is a string, while `age` is an integer.

The easy fix:

```python
age = 17
print(f"Light's age is {age}.")
```

### 1.5 Indexing: Every Character Has a Position

### 1.6 Slicing: Extracting Secret Messages

Imagine finding this message:

```python
message = "KIRAISWATCHING"
```

You only need the first four characters. No need to copy the entire string manually.

Use slicing.

```python
message = "KIRAISWATCHING"

print(message[0:4])
```

**Output:**

```text
KIRA
```

The syntax is:

```python
string[start:end]
```

The start index is included. The end index is excluded.

### More examples

```python
message = "KIRAISWATCHING"

print(message[:4])
print(message[4:7])
print(message[7:])
```

**Output:**

```text
KIRA
ISW
ATCHING
```

Let's decode the logic:

- `[:4]` starts from the beginning and stops before index `4`.
- `[4:7]` starts at index `4` and stops before index `7`.
- `[7:]` starts at index `7` and continues to the end.

### One rule worth remembering

```python
name = "LIGHTYAGAMI"

print(name[0:5])
```

Output:

```text
LIGHT
```

The end index isn't included. Python likes clear boundaries, even when Light doesn't.

### 1.7 String Methods: Editing the Evidence

Real-world data isn't always neatly formatted. A message might contain unnecessary spaces, inconsistent capitalization, or text that needs replacing.

Python provides string methods to handle these situations.

## 1.7.1 `.upper()`: Make Everything LOUD

```python
message = "the investigation continues"

print(message.upper())
```

**Output:**

```text
THE INVESTIGATION CONTINUES
```

*Original detective-style line, not dialogue from the anime.*

## 1.7.2 `.lower()`: Keep It Quiet

```python
message = "KIRA IS WATCHING"

print(message.lower())
```

**Output:**

```text
kira is watching
```

Useful when you want to compare text without worrying about uppercase and lowercase differences.

## 1.7.3 `.strip()`: Remove Unwanted Spaces

Suppose someone submits a suspect's name with extra spaces.

```python
suspect = "   Light Yagami   "

clean_name = suspect.strip()

print(clean_name)
```

**Output:**

```text
Light Yagami
```

`.strip()` removes whitespace from the beginning and end. It doesn't remove spaces between words.

## 1.7.4 `.replace()`: Change the Message

```python
message = "Ryuk likes pears"

updated_message = message.replace("pears", "apples")

print(updated_message)
```

**Output:**

```text
Ryuk likes apples
```

Finally, someone got the details right.

## 1.7.5 `.find()`: Locate a Clue

The `.find()` method returns the index where a substring first appears. If it's not found, it returns `-1`.

```python
message = "The Kira investigation has begun"

print(message.find("Kira"))
print(message.find("L"))
```

**Output:**

```text
4
-1
```

The first result is `4` because `"Kira"` begins at index `4`. The second result is `-1` because uppercase `"L"` does not appear in the string.

### Quick reference

| Method | Purpose |
|---|---|
| `.upper()` | Converts text to uppercase |
| `.lower()` | Converts text to lowercase |
| `.strip()` | Removes surrounding whitespace |
| `.replace(old, new)` | Replaces matching text |
| `.find(text)` | Finds a substring's position, or returns `-1` |

### 1.8 The Most Important Rule: Strings Are Immutable

Here's something that might feel strange at first.

Python strings are **immutable**, meaning you cannot modify an individual character directly.

```python
name = "LIGHT"

# This causes a TypeError:
# name[0] = "N"
```

Python won't let you replace `L` with `N` that way.

But you can create a new string.

```python
name = "LIGHT"

updated_name = "N" + name[1:]

print(updated_name)
```

**Output:**

```text
NIGHT
```

We didn't mutate the original string. We created a new one using the new first character and a slice of the old string.

No Death Note required.
