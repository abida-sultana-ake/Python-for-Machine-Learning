### Accessing characters: Indexing & Slicing
```
 P   Y   T   H   O   N
 0   1   2   3   4   5      ← positive index
-6  -5  -4  -3  -2  -1      ← negative index
```
**slicing:** `s[start : stop : step]` — `stop` **not inclusive** (exclusive).

````python
# indexing

word = "L Lawliet"

print(word[0])
print(word[1])
print(word[2])
print(word[3])
print(word[4])
print(word[5])
````

````python
word = "Near"

print(word[-1])
print(word[-2])
````

````python
# slicing
word = "Python"

print(word[0:3])
print(word[2:6])
print(word[:4])
print(word[2:])
print(word[0:6:2])
print(word[0:6:3])
````

````python
# Reverse
word = "Yagami"
print(word[:])
print(word[::1])
print(word[::-1]) # reverse 
````

````python
# Loop
word = "DEATHNOTE"
for a in word:
    print(a, end = " ")
````

````python
# enumerate
for a,b in enumerate(word):
  print(a, b)
````

````python
# another way access by index
for i in range(len(word)):  #access by index
    print(i, word[i])
````

### String Method

````python
# Imagine L receives this message: 
message = "Kira is watching"
print(message)
print(message.upper())
print(message.lower())
````

````python
character = "light yagami"

print(character.title())
````

````python
# .strip() removes whitespace from the beginning and end of a string. It does not remove the space between Light and Yagami.
suspect = "   Light Yagami   "
clean_name = suspect.strip()
print((clean_name))
````

````python
# .replace(): Change the Message; string.replace(old, new)

message = "Ryuk likes pears"
update_message = message.replace("pears", "apples")
print(message)
print(update_message)
````

````python
message = "Kira is watching. Kira is everywhere."
print(message)
print(message.replace("Kira", "L"))
````

````python
# .find(): Locate a Clue L wants to know where the word "Kira" first appears in a message.

message = "The Kira investigation has begun"

print(message.find("Kira"))

# Why 4? Because Python indexes start at 0, and "Kira" begins at index 4.
# What if the text doesn't exist?

message = "L is investigating"

print(message.find("Kira"))
#The .find() method returns -1 when the requested text is not found. Remember: .find() tells you where text appears. It does not return True or False.
print("L" in message)  # True boolean
````

````python
# .count(): Count the Clues Suppose "Kira" appears multiple times in a case file.

message = "Kira is watching. Kira knows everything."

print(message.count("Kira"))

# The .count() method counts how many times a substring appears. 
message = "Kira and kira and KIRA"

print(message.count("Kira"))
````

````python
# .startswith() and .endswith(): Verify the Message Sometimes an investigator needs to check whether a message begins or ends with particular text.
message = "Kira is watching."

print(message.startswith("Kira"))
print(message.endswith("."))

message = "The investigation continues"
print("___________\n")

print(message.startswith("L"))
print(message.endswith("continues"))
````

````python
#  index() vs find()
print(message.find("python"))
# print(message.index("python")) = title not found error
````

````python
# check methods
print("123".isdigit(), "12a".isdigit())
print("abc".isalpha(), "abc1".isalpha())

print("abc1".isalnum())
print("abc1#".isalnum())

print("hello".islower(), "HELLO".isupper())
print("   ".isspace())
````

````python
# split(): Separate the Evidence
# Imagine L receives a message containing several names in one string:
suspects = "Light L Misa"
suspect_list = suspects.split()

print(suspect_list)

# Split using a comma
characters = "Light,L,Misa,Ryuk"

character_list = characters.split(",")

print(character_list)

#Split a case report
report = "Kira is watching"

words = report.split()

print(words)
print("Number of words:", len(words))
````

````python
# join(): Put the Pieces Back Together
suspects = ["Light", "L", "Misa"]

report = " ".join(suspects)

print(report)

#Join names with commas
characters = ["Light", "L", "Misa", "Ryuk"]

report = ", ".join(characters)

print(report)
````

### f-Strings: Write Reports Like a Detective

````python
# Now L needs to write a report containing a character's name, the case status, and the number of suspects.

# You could combine everything with +, but that can become messy.

# An f-string lets you insert variables and expressions directly into a string.

detective = "L Lawliet"
case_name = "Kira Investigation"
status = "Ongoing"

print(f"Detective: {detective}")
print(f"Case: {case_name}")
print(f"Status: {status}")

# Include calculations
# Ryuk wants to know how many apples he has after finding more.

apples = 5
found_apples = 3

print(f"Ryuk now has {apples + found_apples} apples.")

#Format numbers
case_name = "Kira Investigation"
progress = 75.5678

print(f"Case: {case_name}")
print(f"Investigation progress: {progress:.2f}%")

#Use expressions inside f-strings
light_age = 17
misa_age = 19

print(f"Combined ages: {light_age + misa_age}")
print(f"Age difference: {misa_age - light_age}")
````
