### 1.2 Joining Strings: Connecting the Clues

### 1.3 F-Strings: Making Interrogations Easier


Let's say Light's notebook contains a suspicious entry, and L needs to display a message containing a character's name.

You could use concatenation, but f-strings are often cleaner.

```python
suspect = "Light Yagami"
case = "Kira Investigation"

print(f"Suspect: {suspect}")
print(f"Case: {case}")
```

**Output:**

```text
Suspect: Light Yagami
Case: Kira Investigation
```

An f-string starts with `f` before the quotation mark. Variables inside `{}` are replaced with their values.

Want a more dramatic example?

```python
detective = "L"
case_status = "still suspicious"

print(f"{detective} says the case is {case_status}.")
```

**Output:**

```text
L says the case is still suspicious.
```

*Original Death Note-inspired line. Not a quote from the anime.*

### A classic mistake

```python
age = 17

# This causes a TypeError:
# print("Light's age is " + age)
```

Why? `"Light's age is "` is a string, while `age` is an integer.

The easy fix:

```python
age = 17
print(f"Light's age is {age}.")
```

### 1.5 Indexing: Every Character Has a Position

### 1.6 Slicing: Extracting Secret Messages

Imagine finding this message:

```python
message = "KIRAISWATCHING"
```

You only need the first four characters. No need to copy the entire string manually.

Use slicing.

```python
message = "KIRAISWATCHING"

print(message[0:4])
```

**Output:**

```text
KIRA
```

The syntax is:

```python
string[start:end]
```

The start index is included. The end index is excluded.

### More examples

```python
message = "KIRAISWATCHING"

print(message[:4])
print(message[4:7])
print(message[7:])
```

**Output:**

```text
KIRA
ISW
ATCHING
```

Let's decode the logic:

- `[:4]` starts from the beginning and stops before index `4`.
- `[4:7]` starts at index `4` and stops before index `7`.
- `[7:]` starts at index `7` and continues to the end.

### One rule worth remembering

```python
name = "LIGHTYAGAMI"

print(name[0:5])
```

Output:

```text
LIGHT
```

The end index isn't included. Python likes clear boundaries, even when Light doesn't.

### 1.7 String Methods: Editing the Evidence

Real-world data isn't always neatly formatted. A message might contain unnecessary spaces, inconsistent capitalization, or text that needs replacing.

Python provides string methods to handle these situations.

## 1.7.1 `.upper()`: Make Everything LOUD

```python
message = "the investigation continues"

print(message.upper())
```

**Output:**

```text
THE INVESTIGATION CONTINUES
```

*Original detective-style line, not dialogue from the anime.*

## 1.7.2 `.lower()`: Keep It Quiet

```python
message = "KIRA IS WATCHING"

print(message.lower())
```

**Output:**

```text
kira is watching
```

Useful when you want to compare text without worrying about uppercase and lowercase differences.

## 1.7.3 `.strip()`: Remove Unwanted Spaces

Suppose someone submits a suspect's name with extra spaces.

```python
suspect = "   Light Yagami   "

clean_name = suspect.strip()

print(clean_name)
```

**Output:**

```text
Light Yagami
```

`.strip()` removes whitespace from the beginning and end. It doesn't remove spaces between words.

## 1.7.4 `.replace()`: Change the Message

```python
message = "Ryuk likes pears"

updated_message = message.replace("pears", "apples")

print(updated_message)
```

**Output:**

```text
Ryuk likes apples
```

Finally, someone got the details right.

## 1.7.5 `.find()`: Locate a Clue

The `.find()` method returns the index where a substring first appears. If it's not found, it returns `-1`.

```python
message = "The Kira investigation has begun"

print(message.find("Kira"))
print(message.find("L"))
```

**Output:**

```text
4
-1
```

The first result is `4` because `"Kira"` begins at index `4`. The second result is `-1` because uppercase `"L"` does not appear in the string.

### Quick reference

| Method | Purpose |
|---|---|
| `.upper()` | Converts text to uppercase |
| `.lower()` | Converts text to lowercase |
| `.strip()` | Removes surrounding whitespace |
| `.replace(old, new)` | Replaces matching text |
| `.find(text)` | Finds a substring's position, or returns `-1` |

### 1.8 The Most Important Rule: Strings Are Immutable

Here's something that might feel strange at first.

Python strings are **immutable**, meaning you cannot modify an individual character directly.

```python
name = "LIGHT"

# This causes a TypeError:
# name[0] = "N"
```

Python won't let you replace `L` with `N` that way.

But you can create a new string.

```python
name = "LIGHT"

updated_name = "N" + name[1:]

print(updated_name)
```

**Output:**

```text
NIGHT
```

We didn't mutate the original string. We created a new one using the new first character and a slice of the old string.

No Death Note required.

### 2.9 The Copy Gotcha

**Death Note Story**

L makes a backup of his suspect database. He thinks he can change the backup without affecting the original file. But Python has a little surprise for him!

#### 1. The Problem with Assignment

```python
suspects = ["Light", "Misa", "Ryuk"]

backup = suspects

backup.append("Near")

print(suspects)
print(backup)
```

**Output:**
```text
['Light', 'Misa', 'Ryuk', 'Near']
['Light', 'Misa', 'Ryuk', 'Near']
```

Why did both lists change?

Because `backup = suspects` does **not** create a separate list. Both variables refer to the same list.

#### 2. Making a Real Copy

Use `.copy()` to create a separate copy of the list.

```python
suspects = ["Light", "Misa", "Ryuk"]

backup = suspects.copy()

backup.append("Near")

print(suspects)
print(backup)
```

**Output:**
```text
['Light', 'Misa', 'Ryuk']
['Light', 'Misa', 'Ryuk', 'Near']
```

Now modifying the backup does not change the original list's items.

**Another way to copy a list:**

```python
backup = suspects[:]
```

This also makes a shallow copy of the list.

**Important:** `.copy()` creates a *shallow copy*. If a list contains nested lists, those inner lists are still shared between the original and the copy.



````python
# 1. The Problem with Assignment
suspects = ["Light", "Misa", "Ryuk"]

backup = suspects

backup.append("Near")

print(suspects)
print(backup)

a = """Why did both lists change?
Because backup = suspects does not create a separate list. Both variables refer to the same list."""
print(a)
````

````python
# 2. Making a Real Copy
suspects = ["Light", "Misa", "Ryuk"]

backup = suspects.copy()

backup.append("Near")

print(suspects)
print(backup)
````

### 2.10 List Comprehension

**Death Note Story**

L has a long list of people and wants to create a new list containing only suspects whose names start with `"M"`. Instead of writing a loop across several lines, he uses a compact Python feature called **list comprehension**.

#### 1. Creating a New List

Without list comprehension:

```python
numbers = [1, 2, 3, 4, 5]
squares = []

for number in numbers:
    squares.append(number ** 2)

print(squares)
```

**Output:**
```text
[1, 4, 9, 16, 25]
```

Using list comprehension:

```python
numbers = [1, 2, 3, 4, 5]

squares = [number ** 2 for number in numbers]

print(squares)
```

**Output:**
```text
[1, 4, 9, 16, 25]
```

Both programs produce the same result, but list comprehension is more compact.

#### 2. Selecting Items with a Condition

L only wants suspects whose names start with `"M"`.

```python
suspects = ["Light", "Misa", "Near", "Mello", "Ryuk"]

m_suspects = [name for name in suspects if name.startswith("M")]

print(m_suspects)
```

**Output:**
```text
['Misa', 'Mello']
```

The `if` condition keeps only the names that start with `"M"`.

#### 3. List Comprehension Syntax

```python
new_list = [expression for item in iterable if condition]
```

- `expression`: What to put into the new list.
- `item`: Each item being processed.
- `iterable`: The original collection.
- `if condition`: Optional filter.
---

````python
# 1. Creating a New List

# Without list comprehension:
numbers = [1, 2, 3, 4, 5]
squares = []

for number in numbers:
    squares.append(number ** 2)

print(squares)

# Using list comprehension:
numbers = [1, 2, 3, 4, 5]

squares = [number ** 2 for number in numbers]

print(squares)
````

````python
# 2. Selecting Items with a Condition
suspects = ["Light", "Misa", "Near", "Mello", "Ryuk"]

m_suspects = [name for name in suspects if name.startswith("M")]

print(m_suspects)
````

### Problem 7: Filtering the Numbers Solve

````python
numbers = [3, 8, 12, 5, 20, 7, 15, 2, 18]

greater_numbers = [n for n in numbers if n >= 10]
print(greater_numbers)

even_numbers = [n for n in numbers if n % 2 == 0]
print(even_numbers)
````

### Problem 8: Total and Average Marks Solve

````python
marks = [90, 75, 88, 60, 95]

print(sum(marks))

avg = sum(marks) / len(marks)
print(avg)

print(max(marks))
print(min(marks))

investigators = [n for n in marks if n >= 80]
print(len(investigators))
````

### Problem 9: Duplicate Evidence Solve

````python
evidence = ["apple", "camera", "apple", "notebook", "camera", "fingerprints"]
unique_evidence = []

for item in evidence:
    if item not in unique_evidence:
        unique_evidence.append(item)
print(unique_evidence)
print(len(unique_evidence))
````

### Problem 10: The Final Investigation Report Solve

````python
scores = [45, 78, 92, 34, 88, 67, 100, 55]

# 1. Keep only passing scores (60 or above)
passing_scores = [score for score in scores if score >= 60]

# 2. Sort in descending order
passing_scores.sort(reverse=True)

# 3. Calculate total and average
total = sum(passing_scores)
average = total / len(passing_scores)

# 4. Print the highest passing score
print("Highest passing score:", max(passing_scores))

# 5. Print total, average, and final list
print("Total:", total)
print("Average:", average)
print("Final list:", passing_scores)
````
