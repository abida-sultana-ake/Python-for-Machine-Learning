### 1.3 F-Strings: Making Interrogations Easier


Let's say Light's notebook contains a suspicious entry, and L needs to display a message containing a character's name.

You could use concatenation, but f-strings are often cleaner.

```python
suspect = "Light Yagami"
case = "Kira Investigation"

print(f"Suspect: {suspect}")
print(f"Case: {case}")
```

**Output:**

```text
Suspect: Light Yagami
Case: Kira Investigation
```

An f-string starts with `f` before the quotation mark. Variables inside `{}` are replaced with their values.

Want a more dramatic example?

```python
detective = "L"
case_status = "still suspicious"

print(f"{detective} says the case is {case_status}.")
```

**Output:**

```text
L says the case is still suspicious.
```

*Original Death Note-inspired line. Not a quote from the anime.*

### A classic mistake

```python
age = 17

# This causes a TypeError:
# print("Light's age is " + age)
```

Why? `"Light's age is "` is a string, while `age` is an integer.

The easy fix:

```python
age = 17
print(f"Light's age is {age}.")
```

### 1.6 Slicing: Extracting Secret Messages

Imagine finding this message:

```python
message = "KIRAISWATCHING"
```

You only need the first four characters. No need to copy the entire string manually.

Use slicing.

```python
message = "KIRAISWATCHING"

print(message[0:4])
```

**Output:**

```text
KIRA
```

The syntax is:

```python
string[start:end]
```

The start index is included. The end index is excluded.

### More examples

```python
message = "KIRAISWATCHING"

print(message[:4])
print(message[4:7])
print(message[7:])
```

**Output:**

```text
KIRA
ISW
ATCHING
```

Let's decode the logic:

- `[:4]` starts from the beginning and stops before index `4`.
- `[4:7]` starts at index `4` and stops before index `7`.
- `[7:]` starts at index `7` and continues to the end.

### One rule worth remembering

```python
name = "LIGHTYAGAMI"

print(name[0:5])
```

Output:

```text
LIGHT
```

The end index isn't included. Python likes clear boundaries, even when Light doesn't.

### 1.7 String Methods: Editing the Evidence

Real-world data isn't always neatly formatted. A message might contain unnecessary spaces, inconsistent capitalization, or text that needs replacing.

Python provides string methods to handle these situations.

## 1.7.1 `.upper()`: Make Everything LOUD

```python
message = "the investigation continues"

print(message.upper())
```

**Output:**

```text
THE INVESTIGATION CONTINUES
```

*Original detective-style line, not dialogue from the anime.*

## 1.7.2 `.lower()`: Keep It Quiet

```python
message = "KIRA IS WATCHING"

print(message.lower())
```

**Output:**

```text
kira is watching
```

Useful when you want to compare text without worrying about uppercase and lowercase differences.

## 1.7.3 `.strip()`: Remove Unwanted Spaces

Suppose someone submits a suspect's name with extra spaces.

```python
suspect = "   Light Yagami   "

clean_name = suspect.strip()

print(clean_name)
```

**Output:**

```text
Light Yagami
```

`.strip()` removes whitespace from the beginning and end. It doesn't remove spaces between words.

## 1.7.4 `.replace()`: Change the Message

```python
message = "Ryuk likes pears"

updated_message = message.replace("pears", "apples")

print(updated_message)
```

**Output:**

```text
Ryuk likes apples
```

Finally, someone got the details right.

## 1.7.5 `.find()`: Locate a Clue

The `.find()` method returns the index where a substring first appears. If it's not found, it returns `-1`.

```python
message = "The Kira investigation has begun"

print(message.find("Kira"))
print(message.find("L"))
```

**Output:**

```text
4
-1
```

The first result is `4` because `"Kira"` begins at index `4`. The second result is `-1` because uppercase `"L"` does not appear in the string.

### Quick reference

| Method | Purpose |
|---|---|
| `.upper()` | Converts text to uppercase |
| `.lower()` | Converts text to lowercase |
| `.strip()` | Removes surrounding whitespace |
| `.replace(old, new)` | Replaces matching text |
| `.find(text)` | Finds a substring's position, or returns `-1` |

### 1.8 The Most Important Rule: Strings Are Immutable

Here's something that might feel strange at first.

Python strings are **immutable**, meaning you cannot modify an individual character directly.

```python
name = "LIGHT"

# This causes a TypeError:
# name[0] = "N"
```

Python won't let you replace `L` with `N` that way.

But you can create a new string.

```python
name = "LIGHT"

updated_name = "N" + name[1:]

print(updated_name)
```

**Output:**

```text
NIGHT
```

We didn't mutate the original string. We created a new one using the new first character and a slice of the old string.

No Death Note required.

### 2.9 The Copy Gotcha

**Death Note Story**

L makes a backup of his suspect database. He thinks he can change the backup without affecting the original file. But Python has a little surprise for him!

#### 1. The Problem with Assignment

```python
suspects = ["Light", "Misa", "Ryuk"]

backup = suspects

backup.append("Near")

print(suspects)
print(backup)
```

**Output:**
```text
['Light', 'Misa', 'Ryuk', 'Near']
['Light', 'Misa', 'Ryuk', 'Near']
```

Why did both lists change?

Because `backup = suspects` does **not** create a separate list. Both variables refer to the same list.

#### 2. Making a Real Copy

Use `.copy()` to create a separate copy of the list.

```python
suspects = ["Light", "Misa", "Ryuk"]

backup = suspects.copy()

backup.append("Near")

print(suspects)
print(backup)
```

**Output:**
```text
['Light', 'Misa', 'Ryuk']
['Light', 'Misa', 'Ryuk', 'Near']
```

Now modifying the backup does not change the original list's items.

**Another way to copy a list:**

```python
backup = suspects[:]
```

This also makes a shallow copy of the list.

**Important:** `.copy()` creates a *shallow copy*. If a list contains nested lists, those inner lists are still shared between the original and the copy.



````python
# 1. The Problem with Assignment
suspects = ["Light", "Misa", "Ryuk"]

backup = suspects

backup.append("Near")

print(suspects)
print(backup)

a = """Why did both lists change?
Because backup = suspects does not create a separate list. Both variables refer to the same list."""
print(a)
````

````python
# 2. Making a Real Copy
suspects = ["Light", "Misa", "Ryuk"]

backup = suspects.copy()

backup.append("Near")

print(suspects)
print(backup)
````

### 2.10 List Comprehension

**Death Note Story**

L has a long list of people and wants to create a new list containing only suspects whose names start with `"M"`. Instead of writing a loop across several lines, he uses a compact Python feature called **list comprehension**.

#### 1. Creating a New List

Without list comprehension:

```python
numbers = [1, 2, 3, 4, 5]
squares = []

for number in numbers:
    squares.append(number ** 2)

print(squares)
```

**Output:**
```text
[1, 4, 9, 16, 25]
```

Using list comprehension:

```python
numbers = [1, 2, 3, 4, 5]

squares = [number ** 2 for number in numbers]

print(squares)
```

**Output:**
```text
[1, 4, 9, 16, 25]
```

Both programs produce the same result, but list comprehension is more compact.

#### 2. Selecting Items with a Condition

L only wants suspects whose names start with `"M"`.

```python
suspects = ["Light", "Misa", "Near", "Mello", "Ryuk"]

m_suspects = [name for name in suspects if name.startswith("M")]

print(m_suspects)
```

**Output:**
```text
['Misa', 'Mello']
```

The `if` condition keeps only the names that start with `"M"`.

#### 3. List Comprehension Syntax

```python
new_list = [expression for item in iterable if condition]
```

- `expression`: What to put into the new list.
- `item`: Each item being processed.
- `iterable`: The original collection.
- `if condition`: Optional filter.
---

````python
# 1. Creating a New List

# Without list comprehension:
numbers = [1, 2, 3, 4, 5]
squares = []

for number in numbers:
    squares.append(number ** 2)

print(squares)

# Using list comprehension:
numbers = [1, 2, 3, 4, 5]

squares = [number ** 2 for number in numbers]

print(squares)
````

````python
# 2. Selecting Items with a Condition
suspects = ["Light", "Misa", "Near", "Mello", "Ryuk"]

m_suspects = [name for name in suspects if name.startswith("M")]

print(m_suspects)
````

## 3.5 Using Tuple Methods

L has collected several pieces of evidence related to Kira. Some evidence appears multiple times, and sometimes he needs to find the exact position of a particular item.

Python tuples have two built-in methods that help L investigate these records: `count()` and `index()`.

### 1. The `count()` Method

The `count()` method tells us how many times a particular item appears in a tuple.

**Syntax:**

```python
tuple_name.count(item)
```

Imagine L finds the same type of evidence several times.

```python
evidence = ("Apple", "Notebook", "Apple", "Pen", "Apple")

print(evidence.count("Apple"))
print(evidence.count("Notebook"))
print(evidence.count("Camera"))
```

**Output:**

```text
3
1
0
```

Explanation:

- `"Apple"` appears 3 times.
- `"Notebook"` appears once.
- `"Camera"` does not appear, so the result is `0`.

### 2. The `index()` Method

The `index()` method returns the index of the **first occurrence** of a specified item.

**Syntax:**

```python
tuple_name.index(item)
```

L wants to find the first occurrence of `"Apple"` in the evidence records.

```python
evidence = ("Notebook", "Apple", "Pen", "Apple", "Camera")

print(evidence.index("Apple"))
print(evidence.index("Pen"))
print(evidence.index("Notebook"))
```

**Output:**

```text
1
2
0
```

Remember, Python indexing starts at `0`.

### 3. Finding the First Occurrence of a Repeated Item

What happens if the same item appears multiple times?

```python
evidence = ("Apple", "Pen", "Apple", "Notebook", "Apple")

print(evidence.index("Apple"))
```

**Output:**

```text
0
```

Although `"Apple"` appears three times, `index()` returns only the index of its first occurrence.

You can specify a starting position to search later in the tuple.

```python
evidence = ("Apple", "Pen", "Apple", "Notebook", "Apple")

print(evidence.index("Apple", 1))
```

**Output:**

```text
2
```

Python starts searching from index `1`, so it finds the next `"Apple"` at index `2`.

### 4. What Happens When an Item Does Not Exist?

If `count()` cannot find an item, it returns `0`.

However, if `index()` cannot find an item, Python raises a `ValueError`.

```python
evidence = ("Notebook", "Pen", "Apple")

print(evidence.count("Camera"))
print(evidence.index("Camera"))
```

**Output:**

```text
0
ValueError: tuple.index(x): x not in tuple
```

To avoid this error, check whether the item exists before using `index()`.

```python
evidence = ("Notebook", "Pen", "Apple")

if "Camera" in evidence:
    print(evidence.index("Camera"))
else:
    print("Evidence not found")
```

**Output:**

```text
Evidence not found
```

### 5. Using Both Methods Together

L wants to count how many times `"Apple"` appears and locate its first occurrence.

```python
evidence = ("Apple", "Notebook", "Pen", "Apple", "Camera", "Apple")

apple_count = evidence.count("Apple")
first_apple_index = evidence.index("Apple")

print("Apple Count:", apple_count)
print("First Apple Index:", first_apple_index)
```

**Output:**

```text
Apple Count: 3
First Apple Index: 0
```

```text
Apple Count: 3
First Apple Index: 1
Notebook Count: 1
Camera not found
```

### Key Takeaways

- `count(item)` returns the number of occurrences of an item.
- `index(item)` returns the index of its first occurrence.
- `index(item, start)` starts searching from the specified index.
- `count()` returns `0` when the item is absent.
- `index()` raises a `ValueError` when the item is absent.
- Tuples have only two built-in methods: `count()` and `index()`.

**Next topic: Tuple Operations — checking membership, combining tuples, and repeating their items.**

````python
# The count() Method
evidence = ("Apple", "Notebook", "Apple", "Pen", "Apple")

print(evidence.count("Apple"))
print(evidence.count("Notebook"))
print(evidence.count("Camera"))
````

````python
# 2. The index() Method
evidence = ("Notebook", "Apple", "Pen", "Apple", "Camera")

print(evidence.index("Apple"))
print(evidence.index("Pen"))
print(evidence.index("Notebook"))
````

````python
# 3. Finding the First Occurrence of a Repeated Item
evidence = ("Apple", "Pen", "Apple", "Notebook", "Apple")

print(evidence.index("Apple"))
# Although "Apple" appears three times, index() returns only the index of its first occurrence.

# You can specify a starting position to search later in the tuple.

evidence = ("Apple", "Pen", "Apple", "Notebook", "Apple")

print(evidence.index("Apple", 1))
````

````python
# 4. Using Both Methods Together
evidence = ("Apple", "Notebook", "Pen", "Apple", "Camera", "Apple")

apple_count = evidence.count("Apple")
first_apple_index = evidence.index("Apple")

print("Apple Count:", apple_count)
print("First Apple Index:", first_apple_index)
````

# 3.7 Unpacking Tuples

L has stored important case information inside a tuple. Now he wants to assign each value to a separate variable so he can analyze the evidence more easily.

Python allows us to do this using **tuple unpacking**.

### 1. Basic Tuple Unpacking

Suppose L has a tuple containing a suspect's name, age, and investigation status.

```python
suspect_info = ("Light", 25, "Under Investigation")

name, age, status = suspect_info

print(name)
print(age)
print(status)
```

**Output:**

```text
Light
25
Under Investigation
```

Python assigns each tuple item to the corresponding variable.

- `name` receives `"Light"`.
- `age` receives `25`.
- `status` receives `"Under Investigation"`.

### 2. The Number of Variables Must Match

The number of variables should match the number of tuple items when using ordinary unpacking.

```python
suspects = ("Light", "Misa", "Near")

first, second, third = suspects

print(first, second, third)
```

**Output:**

```text
Light Misa Near
```

But what happens if the number of variables does not match?

```python
suspects = ("Light", "Misa", "Near")

first, second = suspects
```

**Output:**

```text
ValueError: too many values to unpack (expected 2)
```

There are three items but only two variables.

### 3. Using `*` to Collect Remaining Items

```python
suspects = ("Light", "Misa", "Near", "Mello")

print(first)
print(others)
```


Notice that `others` is a **list**, even though the original data is a tuple.

# 3.8 Converting Between Lists and Tuples

The new variable `suspects_tuple` stores the converted data as a tuple.

# 3.9 Working with Nested Tuples

L now has several investigation records. Each record contains a suspect's name and related evidence.

Instead of keeping every record separately, he can store tuples inside another tuple.

This structure is called a **nested tuple**.

### 1. Creating a Nested Tuple

```python
case_files = (
    ("Light", "Notebook"),
    ("Misa", "Phone"),
    ("Near", "Photograph")
)

print(case_files)
```

**Output:**

```text
(('Light', 'Notebook'), ('Misa', 'Phone'), ('Near', 'Photograph'))
```

The outer tuple contains three inner tuples. Each inner tuple stores a suspect and an associated item.

### 2. Accessing Items in a Nested Tuple

Use multiple indexes to access items at different levels.

```python
case_files = (
    ("Light", "Notebook"),
    ("Misa", "Phone"),
    ("Near", "Photograph")
)

print(case_files[0])
print(case_files[1])
```

**Output:**

```text
('Light', 'Notebook')
('Misa', 'Phone')
```

`case_files[0]` returns the first inner tuple. `case_files[1]` returns the second.

### 3. Accessing a Specific Value

To access an individual value inside an inner tuple, use two indexes.

```python
case_files = (
    ("Light", "Notebook"),
    ("Misa", "Phone"),
    ("Near", "Photograph")
)

print(case_files[0][0])
print(case_files[0][1])
print(case_files[2][1])
```

**Output:**

```text
Light
Notebook
Photograph
```

Read `case_files[0][1]` in two steps:

1. `case_files[0]` selects `("Light", "Notebook")`.
2. `[1]` selects `"Notebook"` from that inner tuple.

### 4. Measuring a Nested Tuple

The `len()` function counts items at the level you give it.

```python
case_files = (
    ("Light", "Notebook"),
    ("Misa", "Phone"),
    ("Near", "Photograph")
)

print(len(case_files))
print(len(case_files[0]))
```

**Output:**

```text
3
2
```

The outer tuple contains three records. The first inner tuple contains two items.

### 5. Looping Through a Nested Tuple

L wants to print every suspect and their associated evidence.

```python
case_files = (
    ("Light", "Notebook"),
    ("Misa", "Phone"),
    ("Near", "Photograph")
)

for suspect, evidence in case_files:
    print("Suspect:", suspect)
    print("Evidence:", evidence)
```

**Output:**

```text
Suspect: Light
Evidence: Notebook
Suspect: Misa
Evidence: Phone
Suspect: Near
Evidence: Photograph
```

The `for` loop visits each inner tuple. Tuple unpacking assigns its two items to `suspect` and `evidence`.

## Key Things to Remember

- Dictionaries store **key-value pairs**.
- Keys must be unique.
- Access values using `dictionary[key]` or `get()`.
- Add or update values using `dictionary[key] = value`.
- Delete items using `del`, `pop()`, or `clear()`.
- Use `keys()`, `values()`, and `items()` to iterate.
- Dictionary comprehension creates dictionaries in one expression.

````python
# 5.1 Create a Dictionary
suspects = {
    "Light": 95,
    "Misa": 70,
    "Near": 90
}

print(suspects)
````

````python
# 5.2 Access Dictionary Items
print(suspects["Light"])       # 95
print(suspects.get("Ryuk"))    # None
print(suspects.keys())         # All keys
print(suspects.values())      # All values
````

````python
# 5.3 Add Items
suspects["Mello"] = 85

print(suspects)
````

````python
# 5.4 Update Items
suspects["Misa"] = 80

suspects.update({"Near": 98})

print(suspects)
````

````python
# 5.5 Delete Items
del suspects["Mello"]          # Delete by key
suspects.pop("Misa")           # Remove key and return value
suspects.clear()               # Remove all items
````

````python
# 5.6 Iterate Over a Dictionary
suspects = {"Light": 95, "Misa": 70, "Near": 90}

# Iterate over keys
for name in suspects:
    print(name)

# Iterate over values
for score in suspects.values():
    print(score)

# Iterate over keys and values
for name, score in suspects.items():
    print(name, score)
````

````python
# 5.7 Dictionary Comprehension
scores = {"Light": 95, "Misa": 70, "Near": 90}

# Create a dictionary of passing suspects
passed = {name: score for name, score in scores.items() if score >= 80}

print(passed)
````
