## 1. What Is OOP?

**OOP (Object-Oriented Programming)** is a programming paradigm that organizes code using objects containing data and behavior.

In Python, objects can have:

- **Attributes:** Data associated with an object.
- **Methods:** Functions defined inside a class that describe its behavior.

### Real-Life Example

Think about a student:

- Name: Light Yagami
- Age: 17
- Behavior: Introduce himself

In OOP, we can represent this student as an object.

## 2. What Is a Class?

A **class is a blueprint for creating objects.**

Imagine L creates a standard suspect investigation form. Every suspect will have information organized according to that form.

The form is the class. The individual suspects are the objects.

### Syntax

```python
class ClassName:
    pass
```

- `class` is the keyword used to define a class.
- `ClassName` is the name of the class.
- `pass` is a placeholder that allows the class body to remain empty.

### Example

```python
class Suspect:
    pass
```

We have created a class named `Suspect`.

However, we have not created any individual suspect objects yet.

## 3. What Is an Object?

An **object is an instance of a class.**

Let's create two suspects for L's investigation.

```python
class Suspect:
    pass


light = Suspect()
l = Suspect()

print(light)
print(l)
```

Here:

- `Suspect` is the class.
- `light` is a variable referring to a `Suspect` object.
- `l` is another variable referring to a different `Suspect` object.
- `Suspect()` creates a new instance of the class.

Each call to `Suspect()` creates a separate object.

**Remember: One class can create many objects.**

## 4. What Is a Method?

A **method is a function defined inside a class.**

L needs a way to introduce each suspect in his investigation.

```python
class Suspect:
    def introduce(self):
        print("I am a suspect in the Kira investigation.")


light = Suspect()
light.introduce()
```

### Output

```text
I am a suspect in the Kira investigation.
```

### Understanding the Code

- `def` defines a function.
- `introduce()` is a method of the `Suspect` class.
- `self` refers to the current object.
- `light.introduce()` calls the method using the `light` object.

When we call `light.introduce()`, Python automatically passes `light` as the first argument, which is received by `self`.

**Important:** `self` is a naming convention used for the first parameter of an instance method. It is not a Python keyword.

## 5. One Class, Multiple Objects

Let's create two characters from Death Note.

```python

    def investigate(self):
        print("Investigating the Kira case...")


light = Character()
```

### Output

```text

Investigating the Kira case...
```

Both objects can use the same method because they belong to the same class.

At this stage, both objects behave identically because we have not given them different attributes or customized their behavior.

## 6. Key Differences

| Class | Object |
|---|---|
| A blueprint or template | An instance created from a class |
| Defines attributes and methods | Can hold its own attribute values |
| Example: `Character` | Example: `light` |
| Used to create objects | Used to access attributes and methods |

## 1. Why Do We Need This in AI/ML?

In AI/ML, we often create objects that store information such as:

- Prediction settings
- Training status

For example, we might want to create two different ML model configurations.

```python

```

Both objects come from the same class, but each has different settings.

To make this work, we need `__init__()` and instance attributes.

## 2. What Is `__init__()`?

The `__init__()` method initializes an object when it is created.

```python
class Investigator:
    def __init__(self, name):
        self.name = name


l = Investigator("L")

print(l.name)
```

Output:

```text
L
```

### Understanding the Code

- `class Investigator:` defines the class.
- `__init__(self, name)` receives the initial information.
- `self.name = name` stores that information on the object.
- `Investigator("L")` creates an object and initializes it.
- `l.name` accesses the stored value.

**Remember:** `__init__()` is technically an initializer, although it is commonly called a constructor in beginner tutorials.

## 3. Parameter vs. Instance Attribute

Look carefully at this line:

```python
self.name = name
```

The two names have different roles.

| Expression | Meaning |
|---|---|
| `name` | A parameter receiving the supplied value |
| `self.name` | An attribute stored on the current object |

Think of `name` as the information on L's application form and `self.name` as that information saved in the investigation database.

If L forgets the second part, the database remains empty.

```python
class Investigator:
    def __init__(self, name):
        name = name  # Does not create an instance attribute
```

Correct version:

```python
class Investigator:
    def __init__(self, name):
        self.name = name
```

## 4. One Class, Different Objects

Let's build L's investigation team.

```python
class Investigator:
    def __init__(self, name, skill_level):
        self.name = name
        self.skill_level = skill_level

    def investigate(self):
        print(f"{self.name} is investigating Kira.")


l = Investigator("L", 10)
near = Investigator("Near", 8)

l.investigate()
near.investigate()
```

Output:

```text
L is investigating Kira.
Near is investigating Kira.
```

Both objects use the same method, but the method accesses each object's own attributes.

L has skill level 10. Near has skill level 8.

Ryuk has skill level 100 in eating apples, but unfortunately, that is not a supported ML benchmark.

## 5. Updating an Attribute

An object's attribute can be updated after initialization.

```python
l.skill_level = 11

print(l.skill_level)
```

Output:

```text
11
```

Only `l.skill_level` changes. The value stored in `near.skill_level` remains 8.

This separation is useful when different model objects have different settings.

## 6. Mini AI/ML Example: Model Configuration

```python

        self.learning_rate = learning_rate

```

Output:

```text
Model: Death Note Classifier

```

Here, we are only storing configuration values. We are **not training a real machine learning model yet**.

The example demonstrates how a class can organize settings for different models.

## 7. Common Mistakes

### Mistake 1: Forgetting `self`

```python
def __init__(name):
    self.name = name
```

This is incorrect because the initializer is missing its `self` parameter.

### Mistake 2: Missing an Argument

```python
model = MLModel("Kira Detector")
```

This raises a `TypeError` because `learning_rate` is also required.

Correct:

```python
model = MLModel("Kira Detector", 0.01)
```

### Mistake 3: Confusing Class and Object

```python
class MLModel:
    pass
```

`MLModel` is the class, not an individual configured model.

```python
model1 = MLModel()
```

`model1` refers to an object created from that class.

## 1. Instance Variables

Instance variables store information specific to an individual object.

For example, each investigator has a different name and skill level.

```python
class Investigator:
    def __init__(self, name, skill_level):
        self.name = name
        self.skill_level = skill_level


l = Investigator("L", 10)
near = Investigator("Near", 8)

print(l.name)
print(near.name)
```

### Output

```text
L
Near
```

Here, `name` and `skill_level` are instance variables because each object stores its own values.

## 2. Class Variables

Class variables store information associated with the class itself.

```python
class Investigator:
    department = "Kira Investigation Unit"

    def __init__(self, name):
        self.name = name


l = Investigator("L")
near = Investigator("Near")

print(l.department)
print(near.department)
```

### Output

```text
Kira Investigation Unit
Kira Investigation Unit
```

Both objects can access the same class variable, `department`.

## 3. Using Both Together

```python

        self.name = name
        self.skill_level = skill_level
```


The objects have different personal information but can access the same department name.

## 4. Key Differences

| Instance Variables | Class Variables |
|---|---|
| Belong to individual objects | Belong to the class |
| Often initialized using `self` | Usually defined directly inside the class |
| Can have different values for each object | A shared value is available through the class and its instances |
| Example: `self.name` | Example: `department` |

## 4.1 Class Method

A **class method** works with the class itself rather than a specific object.

We use the `@classmethod` decorator to define a class method.

A class method takes `cls` as its first parameter. `cls` refers to the class.

### Example 1: Investigation Team

L and Near belong to the same investigation team.

```python
class Detective:
    organization = "Kira Investigation Team"

    def __init__(self, name):
        self.name = name

    @classmethod
    def show_organization(cls):
        print(cls.organization)


l = Detective("L")
near = Detective("Near")

Detective.show_organization()
```

**Output:**

```text
Kira Investigation Team
```

### Understanding the Code

- `organization` is a class variable.
- `@classmethod` defines a class method.
- `cls` refers to the `Detective` class.
- `cls.organization` accesses the class variable.

A class method can be called using the class name:

```python
Detective.show_organization()
```

It can also be called through an object:

```python
l.show_organization()
```

Both calls work.

### Example 2: Updating a Class Variable

Suppose the investigation team changes its name.

```python
class Detective:
    organization = "Old Investigation Team"

    @classmethod
    def change_organization(cls, new_name):
        cls.organization = new_name


Detective.change_organization("Kira Investigation Team")

print(Detective.organization)
```

**Output:**

```text
Kira Investigation Team
```

Here, the class method updates the class variable through `cls`.

---

## 4.2 Static Method

A **static method** is a method that belongs to a class but does not automatically receive the instance or class as its first argument.

We use the `@staticmethod` decorator to define it.

A static method does not automatically receive `self` or `cls`.

### Example 1: Checking a Case ID

L needs to check whether a case ID is valid.

```python
class Detective:

    @staticmethod
    def is_valid_case_id(case_id):
        return case_id > 0


print(Detective.is_valid_case_id(101))
print(Detective.is_valid_case_id(-5))
```

**Output:**

```text
True
False
```

### Understanding the Code

- `@staticmethod` defines a static method.
- `case_id` is a regular parameter.
- The method returns `True` if the ID is greater than zero.
- Otherwise, it returns `False`.
- The method does not need an object's data or the class's data.

We can call it directly using the class name:

```python
Detective.is_valid_case_id(101)
```

A static method is useful when a function logically belongs to a class but does not need access to `self` or `cls`.

---

## 4.3 Instance Method vs. Class Method vs. Static Method

| Feature | Instance Method | Class Method | Static Method |
|---|---|---|---|
| Decorator | None required | `@classmethod` | `@staticmethod` |
| First parameter | `self` | `cls` | No automatic `self` or `cls` |
| Works directly with | A particular object | The class | Regular arguments |
| Accesses instance data through `self` | Yes | Not automatically | No |
| Accesses class data through `cls` | Can use the class name or `type(self)` | Yes | Not automatically |

### Easy Way to Remember

- `self` → A particular object
- `cls` → The class
- `@staticmethod` → A method that needs neither an automatically supplied object nor an automatically supplied class

---

## 4.4 A Combined Example

```python
class MLModel:
    framework = "Python ML"

    def __init__(self, model_name):

    def show_model(self):

        print(cls.framework)

    @staticmethod

        return 0 < rate <= 1
```

## 4.5 Common Mistakes

### Mistake 1: Forgetting `cls`

Incorrect:

```python
@classmethod
def show_team():
    print("Kira Investigation Team")
```

Correct:

```python
@classmethod
def show_team(cls):
    print("Kira Investigation Team")
```

A class method must accept `cls` as its first parameter by convention.

### Mistake 2: Using `self` in a Static Method

Incorrect:

```python
@staticmethod
def show_name(self):
    print(self.name)
```

A static method does not automatically receive an object. Use an instance method if you need to access `self.name`.

### Mistake 3: Thinking Static Methods Cannot Receive Arguments

Static methods can accept regular parameters.

```python
@staticmethod
def add(a, b):
    return a + b
```

```python
print(Calculator.add(5, 3))
```

The method does not receive `self` or `cls` automatically, but it can receive any regular arguments that it needs.

# Part E: Single Inheritance and `super()`

## 1. What Is Inheritance?

Inheritance allows a child class to reuse attributes and methods from a parent class.

```python
class Detective:
    def investigate(self):
        print("Investigating the Kira case...")


class SuperDetective(Detective):
    def use_skill(self):
        print("Using a special skill...")


l = SuperDetective()
l.investigate()
l.use_skill()
```

**Output:**
```text
Investigating the Kira case...
Using a special skill...
```

- `Detective` → Parent class
- `SuperDetective` → Child class
- `SuperDetective(Detective)` → Inherits from `Detective`

## 2. What Is `super()`?

`super()` allows a child class to call methods from its parent class.

```python
class Detective:
    def __init__(self, name):
        self.name = name


class SuperDetective(Detective):
    def __init__(self, name, skill):
        super().__init__(name)
        self.skill = skill


l = SuperDetective("L", "Deduction")

print(l.name)
print(l.skill)
```

**Output:**
```text
L
Deduction
```

- `super().__init__(name)` calls the parent class's initializer.
- `self.skill = skill` creates an attribute specific to the child class.

## 3. Key Differences

| Concept | Purpose |
|---|---|
| Inheritance | Reuse parent class functionality |
| `super()` | Call parent class methods |

# Part F: Multiple Inheritance and MRO

## 1. Multiple Inheritance

Multiple inheritance allows a child class to inherit from more than one parent class.

```python
class Investigation:
    def investigate(self):
        print("Investigating the Kira case")


class Deduction:
    def solve_clue(self):
        print("Solving the clue")


class Detective(Investigation, Deduction):
    pass


l = Detective()
l.investigate()
l.solve_clue()
```

**Output:**
```text
Investigating the Kira case
Solving the clue
```

- `Investigation` and `Deduction` are parent classes.
- `Detective` is the child class.
- `Detective(Investigation, Deduction)` inherits from both parents.

## 2. Method Resolution Order (MRO)

MRO determines the order in which Python searches for a method.

```python
class A:
    def show(self):
        print("Class A")


class B:
    def show(self):
        print("Class B")


class C(A, B):
    pass


obj = C()
obj.show()

print(C.mro())
```

**Output of `obj.show()`:**
```text
Class A
```

Python searches `C` first, then `A`, then `B`, then `object`. Since `A` comes before `B`, its `show()` method is called.

Use `ClassName.mro()` to inspect the method resolution order.

## 3. Key Differences

| Concept | Meaning |
|---|---|
| Multiple Inheritance | Inheriting from multiple parent classes |
| MRO | The order Python follows when searching for methods |
