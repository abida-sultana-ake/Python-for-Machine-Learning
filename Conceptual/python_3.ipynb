## 1. What Is OOP?

**OOP (Object-Oriented Programming)** is a programming paradigm that organizes code using objects containing data and behavior.

In Python, objects can have:

- **Attributes:** Data associated with an object.
- **Methods:** Functions defined inside a class that describe its behavior.

### Real-Life Example

Think about a student:

- Name: Light Yagami
- Age: 17
- Behavior: Introduce himself

In OOP, we can represent this student as an object.

## 2. What Is a Class?

A **class is a blueprint for creating objects.**

Imagine L creates a standard suspect investigation form. Every suspect will have information organized according to that form.

The form is the class. The individual suspects are the objects.

### Syntax

```python
class ClassName:
    pass
```

- `class` is the keyword used to define a class.
- `ClassName` is the name of the class.
- `pass` is a placeholder that allows the class body to remain empty.

### Example

```python
class Suspect:
    pass
```

We have created a class named `Suspect`.

However, we have not created any individual suspect objects yet.

## 3. What Is an Object?

An **object is an instance of a class.**

Let's create two suspects for L's investigation.

```python
class Suspect:
    pass


light = Suspect()
l = Suspect()

print(light)
print(l)
```

Here:

- `Suspect` is the class.
- `light` is a variable referring to a `Suspect` object.
- `l` is another variable referring to a different `Suspect` object.
- `Suspect()` creates a new instance of the class.

Each call to `Suspect()` creates a separate object.

**Remember: One class can create many objects.**

## 4. What Is a Method?

A **method is a function defined inside a class.**

L needs a way to introduce each suspect in his investigation.

```python
class Suspect:
    def introduce(self):
        print("I am a suspect in the Kira investigation.")


light = Suspect()
light.introduce()
```

### Output

```text
I am a suspect in the Kira investigation.
```

### Understanding the Code

- `def` defines a function.
- `introduce()` is a method of the `Suspect` class.
- `self` refers to the current object.
- `light.introduce()` calls the method using the `light` object.

When we call `light.introduce()`, Python automatically passes `light` as the first argument, which is received by `self`.

**Important:** `self` is a naming convention used for the first parameter of an instance method. It is not a Python keyword.

## 5. One Class, Multiple Objects

Let's create two characters from Death Note.

```python
class Character:
    def investigate(self):
        print("Investigating the Kira case...")


light = Character()
ryuk = Character()

light.investigate()
ryuk.investigate()
```

### Output

```text
Investigating the Kira case...
Investigating the Kira case...
```

Both objects can use the same method because they belong to the same class.

At this stage, both objects behave identically because we have not given them different attributes or customized their behavior.

## 6. Key Differences

| Class | Object |
|---|---|
| A blueprint or template | An instance created from a class |
| Defines attributes and methods | Can hold its own attribute values |
| Example: `Character` | Example: `light` |
| Used to create objects | Used to access attributes and methods |
