## 4. Series and DataFrame

Pandas has two important data structures.

### Series

A Series is a one-dimensional labeled collection of data.

Imagine a shopping list containing only product prices.

### DataFrame

A DataFrame is a two-dimensional table containing rows and columns.

Imagine your shopping list becoming a complete store inventory with product names, prices, quantities, and categories.

| Product | Price | Quantity |
|---|---:|---:|
| Coffee | 250 | 10 |
| Pizza | 400 | 5 |
| Chocolate | 100 | 20 |

Now your shopping list has become a proper business database!

## 5. Creating a DataFrame

We can create a DataFrame from a Python dictionary.

Each dictionary key becomes a column name, and its values become the data inside that column.

For example, a small coffee shop might store its product names, prices, and available quantities in a DataFrame.

## 6. Understanding Indexes

Every row in a DataFrame has an index.

By default, Pandas starts counting rows from `0`.

It's like students receiving roll numbers, except Pandas doesn't need to call your name three times during attendance!

## 7. Inspecting a DataFrame

Before analyzing a dataset, we should understand what we're working with.

Useful attributes and methods include:

| Method / Attribute | Purpose |
|---|---|
| `head()` | Shows the first five rows by default |
| `tail()` | Shows the last five rows by default |
| `shape` | Returns the number of rows and columns |
| `columns` | Shows column names |
| `dtypes` | Shows data types |
| `info()` | Summarizes the DataFrame |
| `describe()` | Summarizes numerical data statistically |

Remember: `shape`, `columns`, and `dtypes` are attributes, while `head()`, `info()`, and `describe()` are methods.

## 8. Selecting Columns

Need only the product prices? No problem!

Pandas lets us select individual columns or multiple columns without copying the entire table.

This is useful when a dataset contains hundreds of columns but we need only a few.
