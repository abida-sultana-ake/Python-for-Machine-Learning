# 2. Taking Input in Python

`print()` talks. `input()` listens.

`input()` always gives you a **string**. Always.

Even if the student types `18`, Python stores `"18"`.

That is like someone saying "I am 18" and you writing it down as text, not as a number. Then you try to do math with a sentence. Chaos.


## The conversion trick

```text
age = int(input("Enter your age: "))
cgpa = float(input("Enter CGPA: "))
```

- `int(...)` = "please treat this as a whole number"
- `float(...)` = "please treat this as a decimal"
- If they type `eighteen`, `int()` will scream. That scream is called a `ValueError`.

Run the next cell and type when the box appears.


````python
# This cell waits for you. Type a name, then press Enter.
your_name = input("What is your name? ")
print("Nice to meet you,", your_name)
print("Python stored it as:", type(your_name))
````

````python
#problem faced
age = input()
print(age)
print(type(age))
````

````python
#fixed
age = int(input())
print (age)
age = age + 5
print (age)
````

````python
# Type a number. Watch how input() still gives a string...
raw_age = input("Enter your age: ")
print("Raw value:", raw_age, "| type:", type(raw_age))

# ...until you convert it.
age = int(raw_age)
print("Now it is a number:", age, "| type:", type(age))
print("In 5 years you will be", age + 5)
print("(Unless you keep saying you are 21 forever.)")
````

## One line vs two lines

Both are fine. Two lines is easier to debug. One line looks fancy.

```text
age = int(input("Enter age: "))     # fancy
```

```text
raw = input("Enter age: ")          # honest
age = int(raw)
```

If the program crashes, the two-line version lets you print `raw` and see what the human actually typed. Humans type wild things.

# 3. Operators in Python

Operators are the verbs of Python.

Variables sit there. Operators *do* things to them.

If variables are ingredients, operators are the cooking.


## Arithmetic operators (the calculator)

| Operator | Meaning | Tiny memory trick |
|----------|---------|-------------------|
| `+` | add | more samosas |
| `-` | subtract | samosas disappeared |
| `*` | multiply | samosas for the whole table |
| `/` | divide | always gives a float, even `10 / 2` is `5.0` |
| `//` | floor divide | "how many full groups?" throws away the leftover |
| `%` | modulus | leftover after sharing. The honest operator. |
| `**` | power | `2 ** 3` is 8. Python doing drama. |


````python
a = 17
b = 5

print("a + b  =", a + b)
print("a - b  =", a - b)
print("a * b  =", a * b)
print("a / b  =", a / b)      # float. Always.
print("a // b =", a // b)     # how many 5s fit into 17? 3
print("a % b  =", a % b)      # leftover 2
print("a ** b =", a ** b)     # 17 to the power 5. Unnecessarily powerful.
````

## Real-life modulus

`%` is the "what is left?" operator.

- Even number? `n % 2 == 0`
- Last digit? `n % 10`
- Sharing 17 chocolates among 5 friends: each gets `17 // 5`, leftover `17 % 5`


````python
chocolates = 17
friends = 5
print("Each friend gets:", chocolates // friends)
print("You secretly keep:", chocolates % friends)

n = 24
print("Is 24 even?", n % 2 == 0)
print("Last digit of 24:", n % 10)
````

## Comparison operators (they answer True or False)

They do not calculate a new number. They ask a yes/no question.

| Operator | Meaning |
|----------|---------|
| `==` | equal?  (two equals. One equal *assigns*. Two equals *asks*.) |
| `!=` | not equal? |
| `>` `<` `>=` `<=` | the usual judging |


````python
my_marks = 72
pass_mark = 40
scholarship = 80

print("Passed?", my_marks >= pass_mark)
print("Got scholarship?", my_marks >= scholarship)
print("Exactly 72?", my_marks == 72)
print("Not 100, sadly?", my_marks != 100)

# Classic trap:
# =  puts a value in a box
# == asks "are these the same?"
# Mix them up and Python will either crash or silently ruin your day.
````

## Assignment shortcuts

`x += 1` means `x = x + 1`

It is Python saying: "I know you are lazy. I respect that."


````python
lives = 3
print("Start:", lives)

lives -= 1
print("Touched a cactus:", lives)

lives += 2
print("Found a heart item:", lives)

score = 10
score *= 2
print("Combo:", score)
````

# 4. Logical Operators, Precedence, Associativity

## Logical operators

These combine True/False values. Like group decisions.

| Operator | Human version | Example |
|----------|---------------|---------|
| `and` | both must be True | parents agreeing |
| `or` | at least one True | friends agreeing on food |
| `not` | flip it | "I am not sleepy" (lying) |

- `and` is strict. One `False` and the whole thing is `False`.
- `or` is chill. One `True` and the whole thing is `True`.
- `not` is the plot twist.


````python
has_id_card = True
has_fees_paid = False
is_exam_day = True

print("Can enter campus?     ", has_id_card and has_fees_paid)
print("Can sit for exam?     ", has_id_card or is_exam_day)
print("Forgot ID?            ", not has_id_card)

marks = 82
attendance = 90
print("Gets bonus?", marks >= 80 and attendance >= 80)
````

## Short-circuit (bonus personality trait)

Python is lazy in a smart way.

- In `A and B`, if `A` is False, Python does not even look at `B`. Why bother?
- In `A or B`, if `A` is True, Python skips `B`. Already won.

It is like: first friend says "I am not going." For `and`, the plan is dead. No need to call the second friend.


````python
print(False and print("This will not print"))
print(True or print("This will also not print"))
print(True and "okay now I bother")
````

## Precedence: who goes first?

Python does not read left to right like a novel. It has a VIP list.

**Rough order (high to low), the ones you actually need:**

1. `()` parentheses  — the boss. Use them. Always.
2. `**` power
3. `* / // %`
4. `+ -`
5. comparisons: `== != > < >= <=`
6. `not`
7. `and`
8. `or`

If you cannot remember the list, you are a normal human. Use parentheses.

```text
marks > 40 and attendance > 75 or is_teacher_mood_good
```

That line is a tiny crime scene. Put brackets.


````python
print("2 + 3 * 4 =", 2 + 3 * 4)          # 14, not 20. * goes first
print("(2 + 3) * 4 =", (2 + 3) * 4)      # 20. Brackets saved you

marks = 35
attendance = 90
teacher_happy = True

# Without brackets, this is: (marks > 40 and attendance > 75) or teacher_happy
print("Passes somehow?", marks > 40 and attendance > 75 or teacher_happy)

# Say what you actually mean:
print("Real pass rule?", (marks > 40 and attendance > 75) or teacher_happy)
print("Strict pass rule?", marks > 40 and (attendance > 75 or teacher_happy))
````

## Associativity: same-level operators, which side?

When two operators have the **same** precedence, associativity decides the direction.

- Most things are **left to right**: `10 - 3 - 2` is `(10 - 3) - 2` = 5, not 9
- `**` is the rebel. It is **right to left**: `2 ** 3 ** 2` is `2 ** (3 ** 2)` = 512, not 64

You will forget this. That is fine. Brackets are free. Pride is expensive.

````python
print("10 - 3 - 2 =", 10 - 3 - 2)           # left to right -> 5
print("(10 - 3) - 2 =", (10 - 3) - 2)
print("10 - (3 - 2) =", 10 - (3 - 2))

print("2 ** 3 ** 2 =", 2 ** 3 ** 2)         # right to left -> 512
print("(2 ** 3) ** 2 =", (2 ** 3) ** 2)     # 64
print("2 ** (3 ** 2) =", 2 ** (3 ** 2))     # 512

# Memory line:
# Subtract like you read. Power like a plot twist from the end.
````

# 5. Conditional Statements in Python

Programs without `if` do the same thing every time. Boring. Like a vending machine that only sells one drink.

`if` lets Python make decisions.

```text
if it is raining:
    take umbrella
else:
    pretend it will not rain
    (it will rain)
```

**Indentation is not decoration.** In Python, spaces *are* the syntax. If your `if` body is not indented, Python will act like you never said it.

4 spaces. Not "a vibe". Four spaces.

## The shape of a decision

```text
if condition:
    do this
elif other_condition:
    do that
else:
    do the leftover case
```

- `if` is required.
- `elif` is optional. You can stack many.
- `else` is optional. It is the "none of the above" bucket.
- Python checks from the top. The **first True** wins. The rest are ignored. It is not a debate club.


````python
weather = "rain"
has_exam = True

if weather == "rain" and has_exam:
    print("Take umbrella AND admit card. This is not a drill.")
elif weather == "rain":
    print("Umbrella. Maybe stay in and watch a lecture at 1.5x.")
elif has_exam:
    print("Admit card. Water bottle. False confidence.")
else:
    print("Free day. Which means you will still open Instagram first.")
````

## Nested if (if inside if)

Use it when one question depends on another.

Do not nest 7 layers. That is how horror movies start.


````python
age = 19
has_nid = True

if age >= 18:
    if has_nid:
        print("You can register.")
    else:
        print("You are adult-shaped, but bring the card.")
else:
    print("Not yet. Enjoy the last peaceful years.")
````

# 6. While Loop in Python

`if` asks once. `while` asks again and again until the answer is False.

```text
while assignment_not_done:
    panic
    open laptop
    watch one video
    panic again
```

A while loop needs 3 things or it will either never run or never stop:

1. a **start** (counter = 1)
2. a **condition** (counter <= 5)
3. a **change** (counter += 1)

Forget the change and you get an **infinite loop**. The program becomes that relative who says "one last story" for 45 minutes.


````python
count = 1
while count <= 5:
    print("Push-up number", count)
    count += 1    # THE CHANGE. Do not forget this.

print("Done. Your arms are fictional but your loop is real.")
````

## Countdown and a password-style loop

`while True` means "keep going until I `break`."

Useful when you do not know how many tries the user needs. Humans never get it on the first try. It is tradition.


````python
n = 5
while n > 0:
    print(n)
    n -= 1
print("Go!")
````

````python
# A tiny guess loop. Secret number is 7.
# You get 3 tries. After that, Python stops being patient.

secret = 7
tries = 0
max_tries = 3

while tries < max_tries:
    guess = int(input("Guess a number from 1 to 10: "))
    tries += 1
    if guess == secret:
        print("Correct. You may have a future in guessing.")
        break
    else:
        print("Nope. Tries left:", max_tries - tries)
else:
    # This else belongs to the WHILE, not the if.
    # It runs only if the loop finished without a break.
    print("Out of tries. The number was", secret)
````

## `while` / `else` (yes, loops can have else)

If the loop ends **normally**, `else` runs.

If you `break` out, `else` is skipped.

Weird at first. Then kind of elegant. Then you forget it exists. Then you see it in an exam.

# 7. For Loop in Python

`while` is "keep going until I say stop."

`for` is "I already know the collection. Visit each item."

If you have a list of names, a string of letters, or a range of numbers, `for` is the friendly option.

```text
for each samosa in the plate:
    eat it
```

You do not check a condition every time. Python hands you the next item. Like a buffet with a tray.


````python
# range(5) -> 0, 1, 2, 3, 4   (5 numbers, starting from 0)
for i in range(5):
    print("Hello number", i)

print("---")

# range(start, stop) -> stop is EXCLUDED. It is always excluded. Accept it.
for i in range(1, 6):
    print("Count", i)
````

## `range` in one breath

- `range(5)` → 0 to 4
- `range(1, 6)` → 1 to 5
- `range(1, 10, 2)` → 1, 3, 5, 7, 9  (the third number is the step)

The end is never included. Python is exclusive. Like a club that says "open till 10" and locks the door at 9:59.


````python
print("Odd numbers under 10:")
for i in range(1, 10, 2):
    print(i)

print("Countdown with range:")
for i in range(5, 0, -1):
    print(i)
````

````python
friends = ["Nayeem", "Saba", "Tanvir", "Anika"]

for name in friends:
    print("Send notes to", name)

# You can also loop over a string. Each letter gets its 0.2 seconds of fame.
word = "Python"
for ch in word:
    print(ch)
````

## `for` vs `while` (when to use which)

| Use `for` when... | Use `while` when... |
|-------------------|---------------------|
| You know the list / the count | You do not know how many times |
| "Do this 10 times" | "Keep asking until the password is right" |
| Looping through students | Waiting for a condition to become False |

If you are writing `for` with a manual counter and a `break` jungle, you probably wanted `while`.


````python
# Sum of numbers from 1 to 10. Classic. Like "Twinkle Twinkle" for programmers.
total = 0
for n in range(1, 11):
    total += n
print("1 to 10 sum:", total)

# Nested loop: a tiny multiplication table
print("\nMini table:")
for i in range(1, 4):
    for j in range(1, 4):
        print(i, "x", j, "=", i * j)
    print("----")
````

# 8. Break and Continue Statements

These are loop remote controls.

- **`break`** = leave the loop. Completely. Go home.
- **`continue`** = skip *this* round, start the next one.

`break` is walking out of class.

`continue` is sitting down but skipping one boring slide. You are still in the room.

````python
print("break demo: stop when we find 3")
for i in range(1, 8):
    if i == 3:
        print("Found 3. I am done. Bye.")
        break
    print("Checking", i)

print("Loop over. We never even looked at 4, 5, 6, 7.")
````

````python
print("continue demo: skip the even numbers")
for i in range(1, 8):
    if i % 2 == 0:
        continue    # skip the rest of this round
    print("Odd one:", i)

print("We stayed in the loop. We just ghosted the evens.")
````

## Tiny wisdom

- `break` is for "I found what I wanted" or "this is hopeless, stop."
- `continue` is for "this item is trash, next please."
- Do not sprinkle them everywhere. A clean condition is prettier than a loop full of emergency exits.
- Infinite loop? Check your `while` condition and your `+=`. Then check again. Then whisper "please" to the kernel and hit stop.
