
# Part A: Beginner Level

## Problem 1: Basic Calculator

**Topic:** `input()`, Type Conversion, Arithmetic Operators

Take two integers `a` and `b`. Print their sum and product.

**Sample Input**
```text
4 6
```

**Sample Output**
```text
Sum: 10
Product: 24
```

---



## Problem 2: Grade Calculator

**Topic:** Conditional Statements (`if`, `elif`, `else`)

Given an integer `marks`, print the grade according to these rules:

- `A+` if marks >= 80
- `A` if marks >= 70
- `B` if marks >= 60
- `Pass` if marks >= 40
- `Fail` otherwise

Check the highest grade first.

**Sample Input**
```text
74
```

**Sample Output**
```text
A
```

---


## Problem 3: Skip Multiples of 9

**Topic:** `for` Loop, `range()`, Modulus Operator, `continue`

Given an integer `N`, print all multiples of 3 from 1 to N, but skip multiples of 9.

**Sample Input**
```text
20
```

**Sample Output**
```text
3 6 12 15
```

---


## Problem 4: Number Counter

**Topic:** Loops, Conditions, Counting, Modulus Operator

You are given `N` integers. Count:

1. Positive numbers
2. Negative numbers
3. Even numbers
4. Odd numbers

Zero is even but neither positive nor negative.

**Sample Input**
```text
7
-4 0 9 -3 10 5 2
```

**Sample Output**
```text
Positive: 4
Negative: 2
Even: 4
Odd: 3
```

---


## Problem 5: Stop at Negative

**Topic:** Loops, Accumulator, `break`

Read space-separated integers. Add the numbers until the first negative number appears. Do not add the negative number or any numbers after it.

**Sample Input**
```text
5 8 10 -1 100
```

**Sample Output**
```text
23
```

---

## Problem 6: Sum of Odd Numbers

**Topic:** `for` Loop, Conditional Statements, Accumulator, `continue`

Given an integer `N`, calculate the sum of all odd numbers from 1 to N.

**Sample Input**
```text
12
```

**Sample Output**
```text
36
```
